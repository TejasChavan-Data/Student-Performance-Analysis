```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/TejasChavan-Data/Student-Performance-Analysis/main/dataset/student_performance.csv"

df = pd.read_csv(url)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Score"
)

plt.title("Study Hours vs Final Score")
plt.xlabel("Study Hours")
plt.ylabel("Final Score")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Attendance",
    bins=8,
    kde=True
)

plt.title("Student Attendance Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Student Performance Correlation")
plt.show()
```
