```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/TejasChavan-Data/Student-Performance-Analysis/main/dataset/student_performance.csv"

df = pd.read_csv(url)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Final_Score",
    s=100
)

plt.title("Study Hours vs Final Score")
plt.xlabel("Daily Study Hours")
plt.ylabel("Final Score")
plt.grid(alpha=0.3)

plt.savefig(
    "study_hours_vs_score.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x="Attendance",
    bins=8,
    kde=True
)

plt.title("Student Attendance Distribution")
plt.xlabel("Attendance Percentage")
plt.ylabel("Number of Students")

plt.savefig(
    "attendance_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.heatmap(
    df.select_dtypes(include="number").corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Student Performance Correlation")

plt.savefig(
    "correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Correlation with Final Score

correlation = df[
    ["Study_Hours", "Attendance",
     "Previous_Score", "Final_Score"]
].corr()

print("Correlation with Final Score:")

print(
    correlation["Final_Score"]
    .sort_values(ascending=False)
)
```

```python
# Average performance analysis

print("Average Study Hours:",
      df["Study_Hours"].mean())

print("Average Attendance:",
      df["Attendance"].mean())

print("Average Previous Score:",
      df["Previous_Score"].mean())

print("Average Final Score:",
      df["Final_Score"].mean())
```
